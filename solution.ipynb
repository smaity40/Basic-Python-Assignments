7. Factorial Using Function 
Write a function factorial(n) that calculates the factorial of a number. 
Restriction: Use a loop. Do not use recursion.

```python
def factorial(n):
    fact=1
    for _ in range(n,0,-1):
        fact*=_
    return fact

factorial(5)
```

8. Count Digits 
Write a function count_digits(n) that counts the number of digits in a 
number. 
Restriction: Use a while loop. 

```python
def count_digits(n):
    if n==0:
        return 1
    count=0
    while n>0:
        count+=1
        n=n//10
    return count

print(f'number of digits is {count_digits(0)}')
print(f'number of digits is {count_digits(1210)}')
```

9. Reverse a Number 
Write a function reverse_number(n) that reverses a given number. 
Example: 
Input  : 12345 
Output : 54321 
Concepts: while, %, //

```python
def reverse_number(n):
    output=0
    while n>0:
        remainder=n%10
        output=(output*10)+remainder
        n = n//10

    return output

print(f'reverse number is {reverse_number(12345)}')


    
```

10. Prime Number 
Write a function check_prime(n) that checks whether a number is prime. 
Restriction: Use a loop and do not use any external library.

```python
def check_prime(n):
    if n<2:
        print('not prime')
        return 
    
    for _ in range(2,n):
        if n%_==0:
            print("not prime")
            return
    print('prime')

check_prime(-1)
check_prime(0)
check_prime(2)
check_prime(4)
```

```python
def remove_spaces(text):
    new_text=''
    for _ in text.split():
        new_text+=_
    return new_text

print(remove_spaces('Python Programming Language'))

```

21. Count Uppercase and Lowercase Characters 
Write a function count_case(text) that counts: 
• uppercase characters  
• lowercase characters  
• digits  
• spaces  
Example: 
Input: Python123 ABC 
Uppercase : 3 
Lowercase : 6 
Digits    
: 3 
Spaces    
: 1 

```python
def count_case(text):
    uppercase_count=0
    lowercase_count=0
    digits_count=0
    spaces_count=0

    for _ in text:
        if _.isupper():
            uppercase_count+=1
        if _.islower():
            lowercase_count+=1
        if _.isdigit():
            digits_count+=1
        if _.isspace():
            spaces_count+=1

    return uppercase_count,lowercase_count,digits_count,spaces_count


#print(count_case('Python123 ABC'))


result = count_case('Python123 ABC')

print(f"Uppercase={result[0]},\n Lowercase={result[1]},\n Digits={result[2]},\n Spaces={result[3]}")

```

29. Star Pattern 
Write a function star_pattern(n) that prints: 
* 
** 
*** 
**** 
***** 
For n = 5. 
Concepts: nested loops. 

```python
def star_pattern(n):
    for _ in range(1,n+1):
        for i in range(_):
            print("*",end='')
        print()

star_pattern(5)
```

```python
def star_pattern_shortcut(n):
    for _ in range(1, n + 1):
        print("*" * _)  # Multiplies the string and prints the whole row at once

star_pattern_shortcut(5)
```

30. Number Pattern 
Write a function number_pattern(n) that prints: 
1 
12 
123 
1234 
12345 
Concepts: nested for loops.

```python
def number_pattern(n):
    for _ in range(1,n+1):
        for __ in range(1,_+1):
            print(__,end='')
        print()

number_pattern(4)
```
